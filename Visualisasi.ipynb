```
# 2. Tambahkan kolom total untuk memudahkan beberapa grafik
df['total'] = df[['jan', 'feb', 'mar', 'apr']].sum(axis=1)

# Atur style seaborn agar lebih rapi
sns.set_theme(style="whitegrid")


# Bar Chart
plt.figure(figsize=(8, 5)) # Ukuran kompak (lebar 8, tinggi 5 inci)

# Urutkan data dari total tertinggi
df_sorted = df.sort_values('total', ascending=False)
sns.barplot(data=df_sorted, x='nama', y='total', palette='viridis')
plt.title('Total Penjualan per Orang', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45) # Miringkan label agar tidak tumpang tindih
plt.tight_layout()
plt.show()


# Pie chart
plt.figure(figsize=(5, 5)) # Ukuran persegi agar pie chart tidak gepeng

# Hitung total penjualan per bulan
total_per_bulan = df[['jan', 'feb', 'mar', 'apr']].sum()
plt.pie(total_per_bulan, labels=total_per_bulan.index, autopct='%1.1f%%',
startangle=140, colors=sns.color_palette('muted'))
plt.title('Proporsi Penjualan per Bulan', fontsize=12)
plt.tight_layout()
plt.show()


# 3. Stack Plot (Stacked Bar Chart)
# Menampilkan komposisi penjualan per bulan untuk setiap orang
plt.figure(figsize=(8, 5))

# Set 'nama' sebagai index, lalu plot sebagai stacked bar
df.set_index('nama')[['jan', 'feb', 'mar', 'apr']].plot(
    kind='bar',
    stacked=True,
    colormap='Pastel1',
    ax=plt.gca() # Gunakan axis saat ini
    )

plt.title('Komposisi Penjualan Bulanan per Orang (Stacked)', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Jumlah Penjualan')
plt.xticks(rotation=45)
plt.legend(title='Bulan')
plt.tight_layout()
plt.show()


# Histogran
plt.figure(figsize=(6, 4))

sns.histplot(df['total'], bins=5, kde=True, color='teal') # bins=5 membagi

plt.title('Distribusi Frekuensi Total Penjualan', fontsize=12)
plt.xlabel('Range Total Penjualan')
plt.ylabel('Jumlah Karyawan (Frekuensi)')
plt.tight_layout()
plt.show()
```

```
# 6. Time Series Plot
plt.figure(figsize=(7, 4))

# Hitung rata-rata penjualan tim per bulan
rata_bulanan = df[['jan', 'feb', 'mar', 'apr']].mean()

sns.lineplot(x=rata_bulanan.index, y=rata_bulanan.values, marker='o', linewidth='12')
plt.title('Tren Rata-rata Penjualan Tim (Time Series)', fontsize=12)
plt.xlabel('Bulan')
plt.ylabel('Rata-rata Penjualan')
plt.tight_layout()
plt.show()


# 7. Subplot
# Membuat 1 baris, 2 kolom. figsize=(10, 4) agar tetap kompak secara keselur
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
fig.suptitle('Dashboard Analisis Sederhana', fontsize=14, fontweight='bold')

# Subplot 1: Scatter (Kiri)
sns.scatterplot(data=df, x='mar', y='apr', s=80, color='purple', ax=axes[0])
axes[0].set_title('Korelasi Maret vs April')
axes[0].set_xlabel('Maret')
axes[0].set_ylabel('April')

# Subplot 2: Histogram (Kanan)
sns.histplot(df['total'], bins=10, kde=True, color='blue', ax=axes[1])
axes[1].set_title('Distribusi Total Penjualan')
axes[1].set_xlabel('Total')
axes[1].set_ylabel('Frekuensi')

plt.tight_layout(rect=[0, 0, 1, 0.95]) # Memberi ruang
plt.show()
```

```
import pandas as pd
import plotly.express as px

# 1. Muat dataset
url = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv"
df = pd.read_csv(url)

# 2. Ubah format data (melt) khusus mengambil bulan 'jan' dan 'mar'
# Ini wajib dilakukan agar Plotly Express bisa membaca kolom 'bulan'
df_plot = df.melt(
    id_vars='nama',
    value_vars=['jan', 'mar'],
    var_name='bulan',
    value_name='penjualan'
    )

# 3. Buat Line Chart dengan SATU BARIH KODE PLOT
fig = px.line(
    df_plot,
    x='nama',
    y='penjualan',
    color='bulan', # Otomatis memberi warna berbeda
    symbol='bulan', # Otomatis memberi bentuk marker
    symbol_sequence=['circle', 'triangle-up'],# Menentukan bentuk: Bulat & Segitiga
    color_discrete_sequence=['royalblue', 'crimson'], # Menentukan warna: Biru
    markers=True, # Menampilkan marker di setiap
    title='<b>Perbandingan Penjualan: Januari vs Maret</b>',
    labels={'penjualan': 'Jumlah Penjualan', 'nama': 'Nama Karyawan', 'bulan': 'Bulan'
    })

# 4. Atur ukuran agar tidak full screen
fig.update_layout(
    width=800, # Lebar dalam pixel
    height=450, # Tinggi dalam pixel
    title_x=0.5, # Judul rata tengah
    hovermode='x unified' # Hover interaktif (melihat nilai Jan & Mar bersamaan)
    )

# 5. Tampilkan
fig.show()
```

```
#8 Fill Between: Area antara Bulan Januari & Februari
# 'fill_between sangat bagus untuk menyoroti selisih atau fluktuasi antara dua set data. Di sini, kita akan mengisi area di antara penjualan Januari dan Februari untuk setiap karyawan.'

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. Persiapan Data
url = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv"
df = pd.read_csv(url)

# Kita butuh index angka untuk sumbu X pada fill_between dan beberapa plot
x_index = np.arange(len(df))

plt.figure(figsize=(8, 4))

# Plot garis Jan dan Feb terlebih dahulu agar terlihat batasnya
plt.plot(x_index, df['jan'], marker='o', color='blue', label='Januari')
plt.plot(x_index, df['feb'], marker='s', color='orange', label='Februari')

# FILL BETWEEN: Mengisi area di antara garis Jan dan Feb
plt.fill_between(
    x_index,
    df['jan'],
    df['feb'],
    color='gray',
    alpha=0.3, # Transparansi agar garis di belakangnya tetap terlihat
    label='Selisih Jan-Feb'
    )

plt.title('Fluktuasi Penjualan: Januari vs Februari (Fill Between)', fontsize=12, fontweight='bold')
plt.xlabel('Indeks Karyawan')
plt.ylabel('Jumlah Penjualan')
plt.xticks(x_index, df['nama'], rotation=45) # Ganti angka index dengan nama
plt.legend()
plt.tight_layout()
plt.show()
```

```
# 9 Scatter Plot: Semua Bulan
plt.figure(figsize=(8, 5))

# Ubah data menjadi format panjang (long format)
df_melted = df.melt(id_vars='nama', value_vars=['jan', 'feb', 'mar', 'apr'],
                    var_name='Bulan', value_name='Penjualan')

# Buat scatter plot
# Kita gunakan 'Bulan' sebagai sumbu X, dan 'Penjualan' sebagai sumbu Y
scatter = plt.scatter(
    df_melted['Bulan'],
    df_melted['Penjualan'],
    c=df_melted['Penjualan'], # Warna berdasarkan tinggi penjualan (gradient)
    cmap='viridis',           # Palet warna
    s=80,                     # Ukuran titik
    alpha=0.7,                # Sedikit transparan
    edgecolor='black'
    )

plt.title('Scatter Plot Penjualan Semua Bulan', fontsize=12)
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.colorbar(scatter, label='Nilai Penjualan') # Menambahkan legenda warna
plt.tight_layout()
plt.show()
```

```
# Scatter Plot: Hanya 2 Bulan (Korelasi Jan vs Feb)
plt.figure(figsize=(6, 6)) # Ukuran persegi agar proporsional

# Scatter plot dasar
plt.scatter(df['jan'], df['feb'], s=100, color='crimson', alpha=0.7)

# Menambahkan label 'nama' di sebelah setiap titik scatter
for i in range(len(df)):
  plt.annotate(
      df['nama'].iloc[i],
       (df['jan'].iloc[i], df['feb'].iloc[i]),
      xytext=(5, 5), # Geser teks 5 pixel ke kanan dan atas atextcoords='offset points',
      fontsize=9,
      fontweight='bold'
      )

# Menambahkan garis diagonal (y=x) sebagai referensi
# Jika titik di atas garis ini, berarti Feb > Jan. Jika di bawah, Jan > Feb.
max_val = max(df['jan'].max(), df['feb'].max())
plt.plot([0, max_val], [0, max_val], linestyle='--', color='gray', alpha=0.5)

plt.title('Korelasi Penjualan: Januari vs Februari', fontsize=12)
plt.xlabel('Penjualan Januari')
plt.ylabel('Penjualan Februari')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Muat data
url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
df = pd.read_csv(url)

# 2. Bersihkan nama kolom dari spasi tersembunyi (PENTING!)
df.columns = df.columns.str.strip()

# 3. Atur tema visual Seaborn agar terlihat profesional
sns.set_theme(style="whitegrid", font_scale=1.1, palette="muted")

# Distribusi & Hubungan Sederhana
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='KEMISKINAN', bins=30, kde=True, color='coral')

plt.title('Distribusi Tingkat Kemiskinan di Seluruh Daerah', fontsize=12)
plt.xlabel('Tingkat Kemiskinan (%)')
plt.ylabel('Frekuensi (Jumlah Daerah)')
plt.tight_layout()
plt.show()
```

```
# Boxplot (Sebaran IPM & Outlier)
# Melihat median, kuartil, dan daerah mana yang menjadi outlier (pencilan)
# Cara membaut dan menafsirkan Boxplot

plt.figure(figsize=(6, 5))
sns.boxplot(data=df, y='IPM', color='skyblue', width=0.4)

plt.title('Sebaran Indeks Pembangunan Manusia (IPM)', fontsize=12)
plt.ylabel('Nilai IPM (0-100)')
plt.tight_layout()
plt.show()
```

```
# Point Plot (Tren Rata-rata per Tahun)
plt.figure(figsize=(10, 5))

# Menghitung rata-rata kemiskinan per tahun dan memplotnya dengan titik
sns.pointplot(data=df, x='Tahun', y='KEMISKINAN', errorbar=('ci', 95),
              color='crimson', marker='D', linestyles='-')

plt.title('Tren Rata-rata Tingkat Kemiskinan per Tahun', fontsize=12, fontweight='bold')
plt.xlabel('Tahun')
plt.ylabel('Rata-rata Kemiskinan (%)')
plt.xticks(rotation=45) # Miringkan label tahun agar tidak bertumpuk
plt.tight_layout()
plt.show()
```

```
### 3. Tingkat Advanced (Statistical & Dashboard-Style)
plt.figure(figsize=(10, 8))

# Pilih hanya kolom numerik yang relevan untuk korelasi
kolom_korelasi = ['PAD', 'KEMISKINAN', 'IPM', 'bm_pendidikan', 'bm_kesehatan', 'PENGANGGURAN']
df_korelasi = df[kolom_korelasi]

# Hitung matriks korelasi
corr_matrix = df_korelasi.corr()

# Buat mask untuk menyembunyikan segitiga atas (agar tidak redundan)
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Plot heatmap
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt=".2f", cmap='coolwarm',
            linewidths=0.5, cbar_kws={"shrink": .8})

plt.title('Matriks Korelasi Variabel Fiskal & Sosial', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```
# FacetGrid / Relplot (Perbandingan Antar Kelompok)
# Kita buat kategori sederhana berdasarkan PAD untuk contoh ini
df['Kategori_PAD'] = pd.qcut(df['PAD'], q=3, labels=['Rendah', 'Sedang', 'Tinggi'])

plt.figure(figsize=(12, 5))

# relplot membuat grid subplot berdasarkan 'col'
sns.relplot(
    data=df,
    x='bm_pendidikan',
    y='IPM',
    hue='Kategori_PAD',
    col='Kategori_PAD',
    kind='scatter',
    height=4,
    aspect=1.2,
    palette='Set2'
    )

plt.subplots_adjust(top=0.85) # Memberi ruang untuk judul utama
plt.suptitle('Hubungan Belanja Pendidikan dan IPM berdasarkan Kategori PAD')
plt.show()
```
